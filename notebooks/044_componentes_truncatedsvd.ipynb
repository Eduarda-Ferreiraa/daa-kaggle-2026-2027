```python
from pathlib import Path
import sys
import pandas as pd
from IPython.display import display, Image
root = next(p for p in [Path.cwd().resolve(), *Path.cwd().resolve().parents]
            if (p / 'src/ver_resultados.py').is_file())
sys.path.insert(0, str(root))
from src.ver_resultados import carregar_resultado
result = carregar_resultado('044')
print('Resultado real guardado; não foi iniciado novo treino.')
print('Kaggle: por submeter')
selected = result.get('selected_internally', result.get('selected', result.get('selected_unsupervised')))
if isinstance(selected, dict): selected = {k:v for k,v in selected.items() if k not in ['history', 'ae', 'folds', 'age_profile_posthoc', 'sizes']}
print('Configuração selecionada:', selected)
if result.get('paired_comparison'): print('Diferenças emparelhadas nos mesmos folds:', result['paired_comparison'])
metrics = result.get('fixed_metrics', result.get('metrics', {}))
if metrics:
    display(pd.DataFrame([{'parcela':k, 'Macro F1':m['macro_f1'], 'accuracy':m['accuracy']} for k,m in metrics.items() if isinstance(m,dict) and 'macro_f1' in m]))
    if 'validation' in metrics:
        detail = metrics['validation'].get('per_class', metrics['validation'].get('classification_report', {}))
        if detail: display(pd.DataFrame(detail))
rows = result.get('candidates', result.get('results', []))
if rows:
    display(pd.DataFrame([{k:v for k,v in r.items() if k not in ['history','ae','folds','age_profile_posthoc','sizes']} for r in rows]))
if result.get('table'): display(pd.DataFrame(result['table']))
if result.get('examples'): display(pd.DataFrame(result['examples']))
if result.get('error_groups_oof'): display(pd.DataFrame(result['error_groups_oof']))
if result.get('selected_unsupervised'):
    print('Perfil etário posterior; não usado para criar/selecionar clusters')
    display(pd.DataFrame(result['selected_unsupervised'].get('age_profile_posthoc', {})))
for variable, profile in result.get('profiles', {}).items():
    print('Perfil:', variable)
    display(pd.DataFrame(profile))
images = list((root/'reports').glob('experiencia_044_*.png')) + list((root/'reports').glob('044*.png'))
for p in sorted(set(images)): display(Image(filename=str(p)))
print('CSV:', result.get('submission'))
print('Notas:', result.get('note', ''))
```
