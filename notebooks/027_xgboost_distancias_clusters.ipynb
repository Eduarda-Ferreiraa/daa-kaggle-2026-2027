# 027 — XGBoost com distâncias a clusters
- Data: 13/11
**Objetivo:** comparar o XGBoost 007 com features de distâncias a centróides KMeans.

**Alteração face ao controlo:** testar k=4, 8 e 12; representação mista via preparação central, normalização numérica e one-hot categórico ajustados apenas no treino. AGE_RANGE, PID e HHID não constroem clusters.

- As distâncias aos clusters não melhoraram a previsão. A análise dos clusters continua útil para caracterizar os perfis

```python
import pandas as pd
from src.experiments_022_027 import run_experiment, write_final_report

output = run_experiment('027')
result = output['result']
cluster_rows = []
for candidate in result['internal_candidates']:
    details = candidate['cluster_details']
    cluster_rows.append({'variante': 'sem clustering' if details is None else f"k={details['k']}", 'Macro F1 interna': candidate['internal_metrics']['macro_f1'], 'tamanhos': None if details is None else details['sizes'], 'silhouette': None if details is None else details['silhouette_sample'], 'AGE_RANGE médio/mediano (pós-clustering)': None if details is None else details['age_range_description_only']})
display(pd.DataFrame(cluster_rows))
metrics_table = pd.DataFrame(result['fixed_metrics']['per_class'])
display(metrics_table.style.apply(lambda row: ['font-weight: bold' if row['classe'] in [1, 6, 13] else '' for _ in row], axis=1))
print({k: v for k, v in result.items() if k not in ('fixed_metrics', 'internal_candidates', 'prediction_differences')})
write_final_report()
```
